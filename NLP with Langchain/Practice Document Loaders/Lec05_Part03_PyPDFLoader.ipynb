## **3. PyPDFLoader — PDFs Ko Page-by-Page Parhna**

PDFs thori complex hoti hain kyunki unmein multiple pages, images aur formatting hoti hai.  
`PyPDFLoader` har ek **page** ko alag `Document` mein split kar ke yeh kaam aasan bana deta hai.

**Analogy:** Agar TextLoader pori notebook ek sath parhta hai, toh PyPDFLoader har page ko alag kar ke aapko ek ek karke deta hai.

**Sab se behtar hai:** Books, research papers, reports, aur PDF presentations ke liye.

````python
from langchain_community.document_loaders import PyPDFLoader
````

````python
pdf_loader = PyPDFLoader(

    # PDF load karein — har page ek alag Document ban jayega
    file_path="sample.pdf"
)

pdf_docs = pdf_loader.load()

for index , doc in enumerate(pdf_docs):

    print(f"Documents: {index + 1}")
    print("="*90)
    print(f"Content:\n{doc.page_content}")
    print(f"MetaData:\n{doc.metadata}")

print("="*90)
print(f"\nTotal Document: {len(pdf_docs)}")    
````

#### Yahan Kya Hua? (What just happened?)

- `TextLoader` ke muqable mein (jismein humein 1 document mila tha), `PyPDFLoader` ne humein **3 documents** diye — har page ka ek alag document!
- Har document ke metadata mein `page` (page number) aur `source` (file path) shamil hai.
- Page numbers **0** se shuru hote hain, 1 se nahi (bilkul Python ki list indexing ki tarah).

**Mukhtasar Muqabla (Quick Comparison):**

| Loader | Total Documents (#) | Kyun? |
|--------|---------------------|-------|
| `TextLoader` | 1 | Poori file = 1 document |
| `PyPDFLoader` | N (Pages ki tadad) | Har ek page = 1 document |

### Sayein — Kya PDF Read Karne Ka Sirf PyPDFLoader Hi Ek Tarika Hai?

**Bilkul Nahi!** Python mein darhaqeeqat bohot saari PDF libraries hain, aur LangChain ke paas har library ke liye ek alag loader hai. Yeh sab ek hi kaam karte hain (PDFs read karna), lekin in sab ki apni apni khususiyaat (strengths) hain. 

Isay gaadiyon ki misal se samjhein — sedan, SUV, aur truck sab aapko A se B tak le jaate hain, lekin aap raste ke hisab se gaadi chunte hain.

| Loader | Underlying Library | Sub Se Behtar Hai | Speed | Accuracy |
|--------|--------------------|-------------------|-------|----------|
| `PyPDFLoader` | `pypdf` | Simple text PDFs, quick start ke liye | Tez (Fast) | Achi (Good) |
| `PyPDFium2Loader` | `pypdfium2` | Speed-critical tasks, bade batches ke liye | Sab se tez (Fastest) | Achi (Good) |
| `PyMuPDFLoader` | `PyMuPDF` / `fitz` | Complex PDFs (tables, images, formatting) | Tez (Fast) | Behtareen (Best) |
| `PDFPlumberLoader` | `pdfplumber` | Un PDFs ke liye jinse **tables** extract karne hon | Dheemi (Medium) | Tables ke liye Behtareen |
| `UnstructuredPDFLoader` | `unstructured` | Scanned PDFs (OCR), bikhre hue layouts | Dheemi (Slow) | Messy PDFs ke liye Achi |
| `AmazonTextractPDFLoader` | AWS Textract | Bade paimane par Scanned docs (Cloud OCR) | Dheemi (Slow) | Behtareen (Paid) |

### Kab Konsa Loader Istemal Karein?


```
Kya aapki PDF sirf plain text hai?
└─ HAAN → PyPDFLoader (Sub se aasan, jo humne aaj use kiya)

Kya ismein complex tables hain?
└─ HAAN → PDFPlumberLoader ya PyMuPDFLoader

Kya yeh scanned image hai (text select nahi ho sakta)?
└─ HAAN → UnstructuredPDFLoader (OCR ke sath) ya AmazonTextractPDFLoader

1000+ PDFs ke liye maximum speed chahiye?
└─ HAAN → PyPDFium2Loader

Urdu, Arabic, Chinese, Hindi ya doosri non-Latin scripts par kaam kar rahe hain?
└─ HAAN → PyMuPDFLoader (Sub se behtar Unicode support)
```

> **Mukhya Nuktay (Key Lesson):** Sahi loader ka intikhab AAPKE data par depend karta hai.  
> Hamesha pehle yeh puchein: *Yeh kis qism ki document hai? Konsi zuban hai? Kya ismein tables hain? Kya yeh scanned hai?* Phir wahi loader chunein jo fit baithta ho.

## **6. SitemapLoader — Poori Website Ko Crawl / Load Karna**

Kya aap aik page ke bajaye **poori website** load karna chahte hain?  
`SitemapLoader` website ki `sitemap.xml` file ko read karta hai (jismein poori site ke sare pages ki list hoti hai) aur har page ko aik alag Document ke taur par load karta hai.

**Analogy (Misal):** `WebBaseLoader` kitab ka sirf aik page parhta hai. Jabke `SitemapLoader` pehle **table of contents** (fehrist) parhta hai aur phir har ek chapter ko load karta hai.

**Sab Se Behtar Hai:** Documentation sites, company wikis, ya kisi badi website se poora knowledge base banane ke liye.  
**Sambhal Kar Istemal Karein (Warning):** Badi websites ke liye yeh SLOW ho sakta hai — scope ko mahdood (limit) karne ke liye `filter_urls` ka istemal karein.

```python
# SitemapLoader — Reference Only (class demo ke liye yeh zyaada waqt leta hai)
# Apne ghar par try karne ke liye niche diye gaye code ko uncomment karein aur cell run karein!

# IMPORTANT: Jupyter notebooks pehle se ek "event loop" (async engine) run kar rahi hoti hain.
# SitemapLoader internally asyncio ka istemal karta hai, jo Jupyter ke loop ke sath conflict karta hai.
# nest_asyncio isey fix karta hai taake dono aik sath kaam kar sakein.
# Jupyter Notebook mein SitemapLoader use karne se pehle aapko yeh do lines zaroor chalani hongi.
# Agar install nahi hai toh aik baar yeh run karein

# %pip install nest_asyncio  
# %pip install tqdm

import nest_asyncio
nest_asyncio.apply()

from langchain_community.document_loaders.sitemap import SitemapLoader

# filter_urls mahdood (limit) karta hai ke konsi pages load karni hain (warna yeh SAB KUCH load kar dega)
site_map_loader = SitemapLoader(

    web_path="https://docs.python.org/3/sitemap.xml",
    filter_urls=["https://docs.python.org/3/tutorial"],

)

site_map_docs = site_map_loader.load()

print(f"Content Length = {len(site_map_docs)}")
```

---
Jab aap kisi website ka `sitemap.xml` pass karte hain, toh sitemap mein us website ke tamam (ALL) URLs ki list hoti hai.

`Bina filter_urls ke`: SitemapLoader poori website ke saare pages download karne lag jayega, jisse bohot zyaada time lag jayega aur irrelevant data bhi load ho jayega.
filter_urls ke saath: Yeh loader ko keh deta hai ke "Poori website load mat karo, sirf wahi pages download karo jinke URL mein mera diya hua keyword/pattern majood ho."

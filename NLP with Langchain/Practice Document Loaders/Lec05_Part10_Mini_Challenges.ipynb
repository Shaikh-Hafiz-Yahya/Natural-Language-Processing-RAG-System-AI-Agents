```python
# ============================================================
# Challenge 1: The Text Detective
# ============================================================
# Step 1: Create a .txt file (or use the one in data/sample.txt)
# Step 2: Load it with TextLoader
# Step 3: Print the content and metadata

from langchain_community.document_loaders import TextLoader

text_loader = TextLoader(

    file_path=r"C:\Users\Muhammad Yahya\Downloads\python\HopeToSkills ML\Natural-Language-Processing-RAG-System-AI-Agents\NLP with Langchain\Practice Document Loaders\LoRA_QLoRA.txt",
    encoding="utf-8",

)

text_docs = text_loader.load()

print(f"Content(String/Actual Text):\n{text_docs[0].page_content}")
print(f"MetaData(Label):\n{text_docs[0].metadata}")
```

```python
# ============================================================
# Challenge 2: Metadata Inspector
# ============================================================
# Load sample.pdf and print ONLY the source and page number
# from each document's metadata.
#
# Expected output format:
#   Page 1 -> source: data/sample.pdf
#   Page 2 -> source: data/sample.pdf
#   ...

# Your code here:

from langchain_community.document_loaders import PyPDFLoader

pdf_loader = PyPDFLoader(

    file_path="sample.pdf"

)

pdf_docs = pdf_loader.load()

for index , docs in enumerate(pdf_docs):
    print(f"\npage {index + 1} -> source: {pdf_docs[0].metadata.get('source')}")
```

## 11. Quick Reference: Sahi Loader Ka Intekhab (Choosing the Right Loader)

### Sonehra Usool (The Golden Rule)

> **Loaders ko ratta mat marein — yeh sikhein ke sahi loader DHOONDNA kaise hai.**
>
> LangChain mein 200+ loaders maujood hain aur har mahine mazeed add hote hain. Ratta marne ke bajaye, khud se yeh 3 sawalaat poochein:
>
> 1. **Data ki file ya source kis qism ki hai?** (PDF, CSV, Web, Cloud, Database...)
> 2. **Isme kya khas baat hai?** (Tables hain? Scanned hai? Non-English hai? JavaScript-heavy hai?)
> 3. **Search karein:** `langchain [aapki source type] loader`

### Aam Loaders — Aapka Shuruaati Nuqta (Starting Point)

| Aapka Data Source | Default Loader | Behtar Alternates Kab Dhoondne Hain |
|-------------------|----------------|-------------------------------------|
| `.txt` file | `TextLoader` | Badi files ke liye → Chunking ka sochein |
| `.pdf` (sada text) | `PyPDFLoader` | Tables hon → `PDFPlumberLoader`, Scanned ho → `UnstructuredPDFLoader` |
| `.pdf` (pachida/complex) | `PyMuPDFLoader` | Behtareen overall quality, `pip install pymupdf` ki zaroorat hoti hai |
| `.csv` file | `CSVLoader` | Badi CSVs ke liye → `pandas` + custom loader |
| `.docx` / `.pptx` | `UnstructuredFileLoader` | `pip install unstructured` ki zaroorat hoti hai |
| `.html` file | `UnstructuredHTMLLoader` | Ya aasan parsing ke liye `BSHTMLLoader` |
| Web page | `WebBaseLoader` | JavaScript-heavy ho → `PlaywrightURLLoader` |
| Poori website | `SitemapLoader` | Badi websites ke liye slow ho sakta hai, `filter_urls` ka istemal karein |
| Google Drive | `GoogleDriveLoader` | OAuth setup ki zaroorat hoti hai |
| YouTube video | `YoutubeLoader` | `youtube-transcript-api` ki zaroorat hoti hai |
| Notion | `NotionDirectoryLoader` | API token ya exported data ki zaroorat hoti hai |
| AWS S3 | `S3DirectoryLoader` | AWS credentials ki zaroorat hoti hai |
| JSON file | `JSONLoader` | Specific fields nikalne ke liye `jq_schema` ka istemal karein |
| Markdown | `UnstructuredMarkdownLoader` | Heading structure ko barqarar rakhta hai |
| Email (.eml) | `UnstructuredEmailLoader` | Subject, body, aur attachments nikaalta hai |

### Yeh Sirf File Types Ke Baare Mein Nahi Hai!

Yahi usool **har** category ke loader par laagu hota hai. Jis tarah PDFs ke liye 6+ alag loaders hain, waise hi web scraping ke liye bhi kai options maujood hain:

| Masla (Problem) | Default | Behtar Alternate |
|-----------------|---------|------------------|
| Static web page | `WebBaseLoader` | Behtareen kaam karta hai! |
| JavaScript-rendered page | `WebBaseLoader` (fail ho jata hai) | `PlaywrightURLLoader` ya `SeleniumURLLoader` |
| Khas HTML elements ki zaroorat ho | `WebBaseLoader` (bohot ziada text lata hai) | Tags filter karne ke liye `bs_kwargs` ke saath `WebBaseLoader` |
| Poori site crawl karni ho | `SitemapLoader` | Agar sitemap na ho toh `RecursiveUrlLoader` |

**Zehniyat (Mindset) jo banani hai:** Jab koi loader aapko achhe results na de, toh zabardasti usay istemal mat karein — alternate dhoondhein. LangChain mein zaroor uska koi na koi hal maujood hoga!

## Appendix: Is Notebook Mein Istemal Hone Wale PEP 8 Style Rules

Is poori notebook mein code Python ke official style guide (PEP 8) ko follow karta hai.  
Neeche har istemal hone wale rule ka khulasa (summary) aur upar diye gaye code se misalein (examples) maujood hain.

### Naam Rakhne Ke Tariqay (Naming Conventions)

| Rule | Rule Ka Tareeqa | Is Notebook Se Misal |
|------|-----------------|----------------------|
| Variables & functions | `snake_case` | `text_loader`, `pdf_documents`, `explore_pdf()` |
| Constants | `UPPER_CASE` | `TARGET_URL` (section 9) |
| Classes | `PascalCase` | `Document`, `TextLoader` (LangChain se) |

### Imports Ke Qawaneen (Import Rules)

| Rule ID | Rule | Misal |
|---------|------|-------|
| E401 | Ek line mein sirf ek import | `from langchain_community.document_loaders import TextLoader` |
| E402 | Imports file ya notebook section ke sab se upar hon | Tamam imports har code cell ke top par hain |
| — | Imports ki grouping: stdlib → third-party → local | Hum ne is notebook mein sirf third-party imports istemal kiye hain |

### Khali Jagah Aur Formatting (Whitespace & Formatting)

| Rule ID | Rule | Misal |
|---------|------|-------|
| E225 | Operators ke aas paas spaces hon | `index + 1` (na ke `index+1`) |
| E231 | Commas ke baad space ho | `min(3, len(documents))` |
| E265 | Block comments `# ` se shuru hon (# ke baad space ho) | `# Load the PDF` |
| E303 | Function definitions se pehle do khali lines hon | `explore_pdf()` wala cell dekhein |
| E501 | Line ki zyada se zyada lambai 79 characters ho | Badi strings ko slice kiya gaya hai: `doc.page_content[:300]` |

### Behtareen Mashqain (Best Practices)

| Mashq (Practice) | Kyun? | Misal |
|------------------|-------|-------|
| Formatting ke liye f-strings | `.format()` ya `%s` se zyada saaf suthra hai | `f"Pages: {len(docs)}"` |
| Loops ke liye `enumerate()` | Manual counter variables se bachata hai | `for index, doc in enumerate(docs)` |
| Dictionaries ke liye `.get(key, default)` | Agar key missing ho toh code crash nahi hota | `metadata.get("source", "unknown")` |
| Multi-line mein trailing commas | Git mein diffs ko saaf rakhta hai | Multi-line function calls dekhein |
| Functions ke liye Docstrings | Samjhaata hai ke function kya kaam karta hai | `explore_pdf()` mein docstring maujood hai |
| Pehle hi Guard clauses | Agar input galat ho toh shuru mein hi return kar jayein | `if not documents: return` |
| Wazi (Descriptive) variable names | Code angrezi ki tarah padha jata hai | `pdf_loader` (na ke `pl`), `web_documents` (na ke `wd`) |

## 1. Document Object — Aapka Universal Data Container

### Analogy (Misal)

Aik **Shipping Company** ke bare mein sochein. Chahe aap kitabein bhej rahe hon, electronics, ya kapde, har cheez aik **standardized box** mein **shipping label** ke saath jati hai.

LangChain bhi bilkul isi tarah kaam karta hai:

```
  +----------------------------------+
  |        Document Object           |
  +----------------------------------+
  |  page_content = "actual text"    |  <-- The item inside the box
  |  metadata     = {source, page..} |  <-- The shipping label
  +----------------------------------+
```

- **`page_content`** (string) — File se nikala gaya asal text
- **`metadata`** (dictionary) — Extra information: file ki location, page number, author, waghaira.

**Yeh kyun zaroori hai?** Kyunki har loader same `Document` box output karta hai, aap bina kisi dikkat ke PDF loader ki jagah Web loader istemal kar sakte hain aur aapka pipeline smoothly chalega!

````python
from langchain_core.documents import Document

# Chaliye manually aik Document banate hain taake samajh sakein yeh kaisa dikhta hai
sample_doc = Document(
    page_content="LangChain se AI apps banana aasan hai.",
    metadata={
        "source":"manual_creation",
        "page":0,
        "author":"Hope to Skill",
    }
)

# Humare "box" ke andar ki details print karein
print(f"Content: {sample_doc.page_content}")
print(f"MetaData: {sample_doc.metadata}")
print(f"Type: {type(sample_doc)}")
````

````python
# Aap metadata ke fields ko alag alag access kar sakte hain
# .get() ka istemal ['key'] se zyaada safe hai — agar key maujood na ho toh code crash nahi hota
source = sample_doc.metadata.get("source" , "unknown")
meta_data = sample_doc.metadata.get("page" , -1)
author = sample_doc.metadata.get("author" , "N/A")

print(source)
print(meta_data)
print(author)
````

#### Yahan Kya Hua? (What just happened?)
Humne manually (khud se) text aur metadata ke saath aik Document object banaya.

Real-world practice mein aap kabhi bhi inhein manually nahi banayein ge — loaders yeh kaam aapke liye khud kar dete hain.

Lekin ab aapko achi tarah maloom hai ke aapke loaders jo bhi document banayein ge uske andar kya kya hota hai!

Mukhya Nuktay (Key Takeaway): ["key"] ke muqable mein .get("key", default) ka istemal zyaada safe (mehfooz) hai, kyunki agar koi key maujood na ho toh yeh code ko crash karne ki bajaye default value return karta hai.

## 2. TextLoader — Plain Text Files Ko Parhna

Yeh sab se simple loader hai. Yeh aik `.txt` file ko read karta hai aur use `Document` mein wrap kar deta hai.  
Iska matlab hai: *"Pori file ko text ke aik hi block ke taur par parhna."*

**Sab se behtar hai:** `.txt` files, log files, markdown files aur kisi bhi plain text ke liye.

````python
from langchain_community.document_loaders import TextLoader
````

````python
# Step 1: Loader banayein (isey batayein konsi file parhni hai)
text_loader = TextLoader(

    file_path="sample.txt",
    encoding="utf-8",

)

# Step 2: File ko load karein — yeh Documents ki LIST return karta hai
txt_doc = text_loader.load()

# Step 3: Check karein humein kya mila
print(f"Documents loaded: {len(txt_doc)}")
print(f"\n--- The Text --- OR --- Actual Text ---")
print(txt_doc[0].page_content)
print(f"\n--- The Label (Metadata) ---")
print(txt_doc[0].metadata)
````

#### Yahan Kya Hua? (What just happened?)
TextLoader ne poori file ko read kiya aur use aik Document mein daal diya.

Metadata mein sirf source (file ka path) shamil hai — kyunki aik normal text file mein koi pages ya authors nahi hote.

Dhyan dein ke .load() hamesha aik list return karta hai, chahe file mein sirf 1 hi document kyun na ho.

Yaad Rakhne Kabil Pattern: Har loader bilkul aik hi 2-step process ko follow karta hai:

loader = SomeLoader(source) → docs = loader.load()

## 3. PyPDFLoader — PDFs Ko Page-by-Page Parhna

PDFs thori complex hoti hain kyunki unmein multiple pages, images aur formatting hoti hai.  
`PyPDFLoader` har ek **page** ko alag `Document` mein split kar ke yeh kaam aasan bana deta hai.

**Analogy:** Agar TextLoader pori notebook ek sath parhta hai, toh PyPDFLoader har page ko alag kar ke aapko ek ek karke deta hai.

**Sab se behtar hai:** Books, research papers, reports, aur PDF presentations ke liye.

````python
from langchain_community.document_loaders import PyPDFLoader

# PDF load karein — har page ek alag Document ban jayega
pdf_loader = PyPDFLoader(file_path="sample.pdf")
pdf_doc = pdf_loader.load()


print(f"Total pages loaded: {len(pdf_doc)}")
print("=" * 50)

# Sirf pehle 3 pages par loop chalayein (slicing)
for index , doc in enumerate(pdf_doc[:3]):
    print(f"\n---- page {index + 1} ----")

    print(f"Content (first 200 chars): {doc.page_content[:200]}...")

print(f"\nMetaData: {doc.metadata}")    
````

#### Yahan Kya Hua? (What just happened?)

- `TextLoader` ke muqable mein (jismein humein 1 document mila tha), `PyPDFLoader` ne humein **3 documents** diye — har page ka ek alag document!
- Har document ke metadata mein `page` (page number) aur `source` (file path) shamil hai.
- Page numbers **0** se shuru hote hain, 1 se nahi (bilkul Python ki list indexing ki tarah).

**Mukhtasar Muqabla (Quick Comparison):**

| Loader | Total Documents (#) | Kyun? |
|--------|---------------------|-------|
| `TextLoader` | 1 | Poori file = 1 document |
| `PyPDFLoader` | N (Pages ki tadad) | Har ek page = 1 document |

---

### Sayein — Kya PDF Read Karne Ka Sirf PyPDFLoader Hi Ek Tarika Hai?

**Bilkul Nahi!** Python mein darhaqeeqat bohot saari PDF libraries hain, aur LangChain ke paas har library ke liye ek alag loader hai. Yeh sab ek hi kaam karte hain (PDFs read karna), lekin in sab ki apni apni khususiyaat (strengths) hain. 

Isay gaadiyon ki misal se samjhein — sedan, SUV, aur truck sab aapko A se B tak le jaate hain, lekin aap raste ke hisab se gaadi chunte hain.

| Loader | Underlying Library | Sub Se Behtar Hai | Speed | Accuracy |
|--------|--------------------|-------------------|-------|----------|
| `PyPDFLoader` | `pypdf` | Simple text PDFs, quick start ke liye | Tez (Fast) | Achi (Good) |
| `PyPDFium2Loader` | `pypdfium2` | Speed-critical tasks, bade batches ke liye | Sab se tez (Fastest) | Achi (Good) |
| `PyMuPDFLoader` | `PyMuPDF` / `fitz` | Complex PDFs (tables, images, formatting) | Tez (Fast) | Behtareen (Best) |
| `PDFPlumberLoader` | `pdfplumber` | Un PDFs ke liye jinse **tables** extract karne hon | Dheemi (Medium) | Tables ke liye Behtareen |
| `UnstructuredPDFLoader` | `unstructured` | Scanned PDFs (OCR), bikhre hue layouts | Dheemi (Slow) | Messy PDFs ke liye Achi |
| `AmazonTextractPDFLoader` | AWS Textract | Bade paimane par Scanned docs (Cloud OCR) | Dheemi (Slow) | Behtareen (Paid) |

### Kab Konsa Loader Istemal Karein?


```
Is your PDF just plain text?
  └─ YES → PyPDFLoader (simplest, we used this today)

Does it have complex tables?
  └─ YES → PDFPlumberLoader or PyMuPDFLoader

Is it a scanned image (no selectable text)?
  └─ YES → UnstructuredPDFLoader (with OCR) or AmazonTextractPDFLoader

Need maximum speed for 1000+ PDFs?
  └─ YES → PyPDFium2Loader

Working with Arabic, Chinese, Hindi, or other non-Latin scripts?
  └─ YES → PyMuPDFLoader (best Unicode support)
```

> **Key Lesson:** The right loader depends on YOUR data.
> Always ask: *What kind of document is this? What language? Does it have tables?
> Is it scanned?* Then search for the loader that fits.

## 4. CSVLoader — Spreadsheet ki Rows ko Documents mein Badalna

`CSVLoader` aapki CSV file ki **har row** ko aik alag `Document` mein convert kar deta hai. Columns ke naam text ka hissa ban jaate hain.

**Sab se behtar hai:** Product catalogs, customer lists, inventory data, aur survey results ke liye.

````python
from langchain_community.document_loaders.csv_loader import CSVLoader


# Bookstore product catalog load karein
csv_loader = CSVLoader(
    file_path="sample_products.csv",
    encoding="utf-8",
)

csv_document = csv_loader.load()

print(f"Total rows loaded as documents: {len(csv_document)}")
````

````python
# Pehli 3 rows (products) par loop chalayein
for index , doc in enumerate(csv_document[:3]):
    print(f"\n==== Product {index + 1} ====")
    print(f"Content: {doc.page_content}")
    print(f"MetaData: {doc.metadata}")
    print('=' * 90)
````

#### What just happened?

- Our CSV had 8 rows, so we got **8 Documents** (one per row)
- Each document's `page_content` contains the column names AND values  
  (e.g., `product_name: Python Crash Course`)
- The metadata tells us the `source` file and `row` number

**Notice the difference?** A PDF's metadata has `page`, but a CSV's metadata has `row`.  
Different loaders provide different metadata — always check what you're getting!

## 5. WebBaseLoader — Websites se Data Fetch Karna

Agar aapka data file mein hone ki bajaye kisi **website** par ho, toh `WebBaseLoader` BeautifulSoup ka istemal karke web pages ko scrape karta hai aur unhein Documents mein convert karta hai.

**Sab se behtar hai:** Documentation sites, blog posts, Wikipedia articles, aur news web pages ke liye.

````python
from langchain_community.document_loaders import WebBaseLoader

# Wikipedia ka NLP page load karein
web_page_loader = WebBaseLoader(

    web_path=["https://en.wikipedia.org/wiki/Natural_language_processing"],
)

web_documents = web_page_loader.load()

print(f"Document Loader: {len(web_documents)}")
print(f"\n--- Metadata ---")
print(web_documents[0].metadata)
print(f"\n--- Content (first 500 chars) ---")
print(web_documents[0].page_content[:500])
````

````python
# Aap AI / Web scrapers ke liye aik sath BOHOT SAARI (Multiple) web pages load kar sakte hain!
multi_loader = WebBaseLoader(
    web_path=[

        "https://en.wikipedia.org/wiki/Machine_learning",
        "https://en.wikipedia.org/wiki/Artificial_intelligence",
    ],
)

multi_docs = multi_loader.load()

# Yeh loop 'multi_docs' ke HAR aik document par chalega (slicing ki zaroorat nahi kyunki sirf 2 docs hain)
# enumerate() humein har document ke saath uska index (0, 1) deta hai
for index , doc in enumerate(multi_docs):
    # .get() ka faida yeh hai ke yeh metadata dictionary se key ko bina kisi error ke read kar leta hai
    # Agar key na mile, toh yeh fallback value ("Unknown" / "No Title") return kar deta hai
    soure = doc.metadata.get('source' , 'Unknown')
    title = doc.metadata.get('title' , 'No Title')

    # len() page_content ke andar majood total characters ko count karta hai
    content_length = len(doc.page_content)
    print(f"Documents: {index + 1}")
    print(f"  Title: {title}")
    print(f"  Source: {source}")

    # f-string ke andar :, use karne se bade numbers mein commas aa jate hain (e.g., 85432 → 85,432)
    print(f"  Content Length: {content_length:,} characters")
    print("="*90)
````

#### Yahan Kya Hua? (What just happened?)

- Har URL **aik Document** ban gaya (poore page ka text).
- Web metadata mein `source` (URL), `title`, aur kabhi kabhar `description` aur `language` bhi shamil hoti hain.
- Content shayad thora messy (bikhra hua) dikhayi de — kyunki web pages par navigation menus, footers aur doosra extra text bhi shamil hota hai. Isay saaf (clean) karne ka tariqa hum aage ke lectures mein sikhein ge!

**Tip:** `{content_length:,}` bade numbers mein commas add kar deta hai (maslan `85432` ki bajaye `85,432`). Readability (parhne ke qabil banane) ke liye yeh Python ki aik behtareen trick hai!

## 8. Hands-On Part 1: PDF Explorer

Aayein ab sab kuch aapas mein milate hain! Ek reusable function banate hain jo kisi bhi PDF ko load kare aur humein ek behtareen summary de. Yeh ek aisi cheez hai jo aap apne real-world projects mein amlan istemal karenge.

````python
from langchain_community.document_loaders import PyPDFLoader
````

````python
def explore_pdf(file_path):
    """PDF load karta hai aur uske contents ki aik acchi summary print karta hai."""

    # Har page ko ek alag Document ke taur par load karein
    pypdf_loader = PyPDFLoader(

        file_path=file_path

        )

    documents = pypdf_loader.load()

    # Summary header
    print(f"PDF: {file_path}")
    print(f"Total Pages: {len(documents)}")
    print("=" * 60)  # Visual divider ke taur par 60 equal signs print karta hai

    # min(3, len(documents)) CHHOTI value ko select karta hai
    # Agar PDF ke 50 pages hain toh 3 dikhayega, agar 2 pages hain toh 2 dikhayega.
    # Yeh code ko crash hone se bachata hai jab pages 3 se kam hon.
    page_to_show = min(3 , len(documents))

    # range(pages_to_show) numbers generate karta hai: 0, 1, 2 (pages_to_show - 1 tak)
    # Hum inhein index ke taur par list se har document access karne ke liye use karte hain
    for i in range(page_to_show):
        doc = documents[i]
        print(f"\n------Page {i + 1}-------")

        # [:300] sirf pehle 300 characters dikhata hai preview ke liye
        # POORA page dekhne ke liye is se replace karein: print(doc.page_content)
        print(f"Content preview: {doc.page_content[:300]}")
        print(f"Metadata: {doc.metadata}")
        print("-" * 60)

        # Dikhayein ke metadata mein konsi keys available hain
    if documents:
        # .keys() metadata dictionary ki tamam keys ke naam return karta hai
        # list() isey aik readable list format mein convert karta hai
        print(f"\nAvailable metadata keys: {list(documents[0].metadata.keys())}")

        # Documents return karein taake baad mein code mein istemal ho sakein
    return documents

# Try karein! 'sample.pdf' ko apne kisi PDF path se replace kar sakte hain
pdf_docs = explore_pdf('sample.pdf')
````

#### Yahan Kya Hua? (What just happened?)

Humne ek **reusable function** banaya jo kisi bhi PDF ke sath kaam karta hai. Kuch aham baatein jinhe note karna zaroori hai:

- `min(3, len(documents))` yeh yaqini banata hai ke agar PDF mein 3 se kam pages hon toh code crash na ho.
- `documents[0].metadata.keys()` humein dikhata hai ke loader konsi extra information (metadata) provide kar raha hai.
- Yeh function documents ko **return** karta hai taake aap unhein baad mein apne pipeline (RAG ya Q&A system) mein istemal kar sakein.

# 🔍 Python `hasattr()` Function — Notes & Reference

## 1. What is `hasattr()`? (Yeh Kya Hai?)

`hasattr()` Python ka ek built-in function hai jiska matlab hai **"Has Attribute?"** (Kya is object ke paas yeh property/variable majood hai?).

Yeh check karta hai ke kisi object ke andar koi specific property, variable, ya method majood hai ya nahi aur **`True`** ya **`False`** return karta hai.

---

## 2. Syntax & Parameters

```python
hasattr(object, attribute_name)

ParameterTypeDescriptionobjectObjectJis object ko aap check karna chahte hain (e.g., web_docs[0]).attribute_nameString (str)Us property ya method ka naam quotes ke andar ('page_content', 'metadata').3. Basic Code ExamplePythonclass Car:
    brand = "Toyota"
    wheels = 4

my_car = Car()

# Checks if properties exist

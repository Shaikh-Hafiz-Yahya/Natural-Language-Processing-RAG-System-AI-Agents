## 4. RecursiveCharacterTextSplitter — Smart Pizza Cutter

Yeh LangChain mein **sab se ziada istemal hone wala (#1) splitter** hai, aur aap ke liye sab se behtareen default intekhab hai.

### Isay "Recursive" Kyun Kehte Hain?

Yeh text ko **qudrati sarhadon (natural boundaries)** par todne ki koshish karta hai, aur ek tarjeehi fehrist (priority list) par kaam karta hai:

Priority 1: Paragraph breaks par todein     
(agar chunks abhi bhi boht bade hain)

Priority 2: Line breaks par todein           
(agar chunks abhi bhi boht bade hain)

Priority 3: Spaces par todein                
(agar chunks abhi bhi boht bade hain)

Priority 4: Kisi bhi character par todein

---

**Haqiqi Misal (Analogy):** Farz karein aap se kisi chapter ko chote hisson mein taqseem karne ko kaha jaye.  
Aap pehle paragraphs ke beech mein se taqseem karne ki koshish karenge. Agar koi paragraph abhi bhi boht lamba ho, toh aap sentences (jumlon) ke beech mein se todein ge. Bilkul aakhri charay ke tor par hi aap lafz ke beech mein se katenge.

`RecursiveCharacterTextSplitter` bilkul isi tarah kaam karta hai!

*`Human Brain ki trha Decision lyta ha RecursiveCharacterTextSplitter`*

```python
from langchain_community.document_loaders import TextLoader

txt_loader = TextLoader(

    file_path='data/nlp_article.txt',
    encoding='utf-8'

)

txt_docs = txt_loader.load()

full_txt = txt_docs[0].page_content
print(f'{'_'*100}\nActualText(String):\n{'_'*100}\n{full_txt}')

meta_data = txt_docs[0].metadata
print(f'{'_'*100}\nMetaData(Label/Dictionary):\n{'_'*100}\n{'_'*4}{meta_data}{'_'*4}')

print('_'*100)
print(f'Total Characters are = {len(txt_docs[0].page_content):,}')
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# Apni pasandida settings ke sath splitter create karte hain
# chunk_size = har chunk mein ziada se ziada characters ki tadad
# chunk_overlap = chunks ke beech dohraaye jaane wale characters ki tadad
splitter = RecursiveCharacterTextSplitter(

    chunk_size=1000,     # har chunk ziada se ziada 1000 characters ka hoga
    chunk_overlap=200,   # 200 chars ka overlap (20% — bilkul munasib hissa)

)

# Apne loaded document ko chunks mein split karte hain
# .split_documents() Documents ki list leta hai aur chote Documents ki list return karta hai
chunks = splitter.split_documents(documents=txt_docs)

print(f"Asal document: 1 document, {len(full_txt):,} characters")
print(f"Splitting ke baad: {len(chunks)} chunks")
print(f"\nSettings: chunk_size=1000, chunk_overlap=200")
```

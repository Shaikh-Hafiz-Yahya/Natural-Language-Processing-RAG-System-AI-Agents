## 5. Doosre Splitter Types — Khas Kaamon Ke Liye Khas Tools

**`RecursiveCharacterTextSplitter` aapka default splitter hai.** Lekin jaise Lecture 5 mein humne dekha tha ke alag alag PDFs ke liye alag loaders hote hain, bilkul waise hi alag *content types* ke liye alag splitters ki zaroorat hoti hai.

| Splitter | Kiske Liye Behtareen Hai | Yeh Kaise Split Karta Hai |
|----------|---------------------------|---------------------------|
| `RecursiveCharacterTextSplitter` | Aam text (aapka default) | Paragraphs → lines → spaces → chars |
| `CharacterTextSplitter` | Sada aur teez splitting | Ek hi separator par todtay hain (jaise `\n\n`) |
| `TokenTextSplitter` | Jab aapko exact token counts chahiye hon | Characters ki bajaye tokens ke hisab se todtay hain |
| `MarkdownTextSplitter` | Markdown docs, wikis | `#` headers par todtay hain — structure barqarar rehta hai |
| `RecursiveCharacterTextSplitter.from_language()` | Source code / Programming files | Function aur Class ki sarhadon (boundaries) par todtay hain |
| `HTMLSectionSplitter` | HTML web pages | HTML tags par todtay hain jaise `<h1>`, `<h2>` |

# ============================================================
# CharacterTextSplitter — Sadi Aur Aasan Cutter
# ============================================================
Yeh sirf EK SINGLE separator par text todta hai. Teez hai, lekin Recursive jitna smart nahi.
Kiske liye acha hai: Aisa data jisme pehle se saaf separators hon (jaise logs ya CSV-style text)

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter
```

```python
txt_loader = TextLoader(

    file_path="data/nlp_article.txt",
    encoding="utf-8",

)

load_docs = txt_loader.load()

print(load_docs[0].page_content)# Actual Text 
print('-'*100)
print(load_docs[0].metadata)
print('-'*100)
```

```python
from langchain_text_splitters import CharacterTextSplitter

# Sirf double newlines (paragraph breaks) par text todte hain
simple_splitter = CharacterTextSplitter(

    separator="\n\n",    # sirf paragraph breaks par todega — paragraph ke beech mein nahi katenge
    chunk_size=1000,
    chunk_overlap=200,

)

simple_chunks = simple_splitter.split_documents(documents=load_docs)

print(f"CharacterTextSplitter se bane: {len(simple_chunks)} chunks")

# [:300] pehle chunk ka preview dikhata hai (sirf pehle 300 characters)
# Poora pehla chunk dekhne ke liye istemal karein: print(simple_chunks[0].page_content)
print(f"\nPehla chunk ({len(simple_chunks[0].page_content)} chars):")
print(simple_chunks[0].page_content[:300])
print(f"{'_'*100}\nComplete First Chunk(773 chars):\n{simple_chunks[0].page_content}")
```

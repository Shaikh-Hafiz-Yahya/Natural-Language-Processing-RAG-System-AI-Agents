## 5. WebBaseLoader — Websites se Data Fetch Karna

Agar aapka data file mein hone ki bajaye kisi **website** par ho, toh `WebBaseLoader` BeautifulSoup ka istemal karke web pages ko scrape karta hai aur unhein Documents mein convert karta hai.

**Sab se behtar hai:** Documentation sites, blog posts, Wikipedia articles, aur news web pages ke liye.

```python
from langchain_community.document_loaders import WebBaseLoader
```

```python
# Wikipedia ka NLP page load karein
web_base_loader = WebBaseLoader(

    web_path=["https://en.wikipedia.org/wiki/Natural_language_processing"],

)

web_docs = web_base_loader.load()

print(f"Document Loader: {len(web_docs)}")
print(f"MetaData/Label\n{web_docs[0].metadata}")
print(f"Content/(String)\n{web_docs[0].page_content[:100]}")
```

```python
print(f"----MetaData----\ntitle: {web_docs[0].metadata.get('title')}")
print(f"source: {web_docs[0].metadata.get('source')}")
```

```python
# Aap AI / Web scrapers ke liye aik sath BOHOT SAARI (Multiple) web pages load kar sakte hain!

multi_web_loader = WebBaseLoader(

    web_path=[

        "https://en.wikipedia.org/wiki/Machine_learning",
        "https://en.wikipedia.org/wiki/Artificial_intelligence"

        ],

)

multi_web_docs = multi_web_loader.load()
print(f"==============MultiWebDocuments Are:{len(multi_web_docs)}==============")

# Yeh loop 'multi_web_docs' ke HAR aik document par chalega (slicing ki zaroorat nahi kyunki sirf 2 docs hain)
# enumerate() humein har document ke saath uska index (0, 1) deta hai
for index , doc in enumerate(multi_web_docs):

    print(f"Index :\n{index}")
    print(f"{doc.page_content[:10]}")
    print(f"MetaData: {doc.metadata}")
    print("="*100)


print(f"\nTotal Documents Are:{index + 1}")
print(f"Document#01 MetaData:\ntitle : {multi_web_docs[0].metadata.get("title")}")
print(f"Document#02 MetaData:\ntitle : {multi_web_docs[1].metadata.get("title")}")
```

```python
print(f"Content: {multi_web_docs[0].page_content[:200]}")
```

```python
print(f"Content: {multi_web_docs[1].page_content[:200]}")
```

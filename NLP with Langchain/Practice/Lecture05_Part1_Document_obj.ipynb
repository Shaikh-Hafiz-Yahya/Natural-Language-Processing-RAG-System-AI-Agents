# **`Part01` Document Object**

```python
from langchain_core.documents import Document
```

```python
sample_docs = Document(

    page_content="Langchain sy AI APP build krna is very easy",
    metadata={

        "source" : "manual_creation",
        "page" : 0,
        "author" : "Muhammad Yahya",

    }

)

print(f"---File se nikala gaya asal text---\nContent : {sample_docs.page_content}")
print(f"\n---Extra information:file ki location, page number, author, waghaira---\nMetaData : {sample_docs.metadata}")
print(f"\nType : {type(sample_docs)}")
```

```python
# Aap metadata ke fields ko alag alag access kar sakte hain
# .get() ka istemal ['key'] se zyaada safe hai — agar key maujood na ho toh code crash nahi hota
source = sample_docs.metadata.get("source" , "Unknown")
page = sample_docs.metadata.get("page" , -1)
author = sample_docs.metadata.get("author" , "N/A")

print(f"source : {source}")
print(f"page : {page}")
print(f"author : {author}")
```
